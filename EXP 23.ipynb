```python
# Alpha-Beta Pruning Algorithm

def minimax(depth, node, maximizing, values, alpha, beta):

    # Leaf node
    if depth == 3:
        return values[node]

    if maximizing:
        best = -999

        for i in range(2):
            value = minimax(depth + 1, node * 2 + i,
                            False, values, alpha, beta)

            best = max(best, value)
            alpha = max(alpha, best)

            # Beta Pruning
            if beta <= alpha:
                break

        return best

    else:
        best = 999

        for i in range(2):
            value = minimax(depth + 1, node * 2 + i,
                            True, values, alpha, beta)

            best = min(best, value)
            beta = min(beta, best)

            # Alpha Pruning
            if beta <= alpha:
                break

        return best


# Leaf node values
values = [3, 5, 6, 9, 1, 2, 0, -1]

alpha = -999
beta = 999

result = minimax(0, 0, True, values, alpha, beta)

print("Optimal value:", result)
```
