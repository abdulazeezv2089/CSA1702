```python
# Set Operations in Python

A = {1, 2, 3, 4, 5}
B = {4, 5, 6, 7, 8}

print("Set A:", A)
print("Set B:", B)

# 1. Union
print("Union:", A | B)

# 2. Intersection
print("Intersection:", A & B)

# 3. Difference
print("A - B:", A - B)
print("B - A:", B - A)

# 4. Symmetric Difference
print("Symmetric Difference:", A ^ B)

# 5. Subset
print("Is A subset of B?", A.issubset(B))

# 6. Superset
print("Is A superset of B?", A.issuperset(B))
```
