```python
# Transpose of a Matrix

A = [
    [1, 2, 3],
    [4, 5, 6]
]

transpose = [
    [A[j][i] for j in range(2)]
    for i in range(3)
]

print("Original Matrix:")
for row in A:
    print(row)

print("Transpose Matrix:")
for row in transpose:
    print(row)
```
