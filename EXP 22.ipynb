```python
# Minimax Algorithm for Tic-Tac-Toe

board = [" "] * 9

def check_winner(p):
    wins = [
        (0,1,2), (3,4,5), (6,7,8),
        (0,3,6), (1,4,7), (2,5,8),
        (0,4,8), (2,4,6)
    ]
    return any(board[a] == board[b] == board[c] == p
               for a, b, c in wins)

def minimax(maximizing):
    if check_winner("O"):
        return 1
    if check_winner("X"):
        return -1
    if " " not in board:
        return 0

    if maximizing:
        best = -10
        for i in range(9):
            if board[i] == " ":
                board[i] = "O"
                best = max(best, minimax(False))
                board[i] = " "
        return best
    else:
        best = 10
        for i in range(9):
            if board[i] == " ":
                board[i] = "X"
                best = min(best, minimax(True))
                board[i] = " "
        return best

def best_move():
    score = -10
    move = 0

    for i in range(9):
        if board[i] == " ":
            board[i] = "O"
            value = minimax(False)
            board[i] = " "

            if value > score:
                score = value
                move = i

    return move

# Example game state
board = ["X", "O", "X",
         " ", "O", " ",
         " ", " ", "X"]

print("Current Board:")
for i in range(0, 9, 3):
    print(board[i], "|", board[i+1], "|", board[i+2])

move = best_move()

print("\nBest move for Computer (O):", move + 1)
```
