```python
import heapq

# Graph with costs
graph = {
    'A': [('B', 1), ('C', 4)],
    'B': [('D', 2), ('E', 5)],
    'C': [('D', 1)],
    'D': [('E', 3)],
    'E': []
}

# Heuristic values
h = {
    'A': 6,
    'B': 4,
    'C': 2,
    'D': 3,
    'E': 0
}

def a_star(start, goal):
    open_list = []
    heapq.heappush(open_list, (h[start], 0, start, [start]))

    visited = set()

    while open_list:
        f, g, node, path = heapq.heappop(open_list)

        if node in visited:
            continue

        visited.add(node)

        if node == goal:
            return path, g

        for neighbour, cost in graph[node]:
            if neighbour not in visited:
                new_g = g + cost
                new_f = new_g + h[neighbour]

                heapq.heappush(
                    open_list,
                    (new_f, new_g, neighbour, path + [neighbour])
                )

    return None, float('inf')


# Start and goal
start = 'A'
goal = 'E'

path, cost = a_star(start, goal)

print("A* Path:", " -> ".join(path))
print("Total Cost:", cost)
```
