```python
# Add Two Matrices

A = [
    [1, 2, 3],
    [4, 5, 6]
]

B = [
    [7, 8, 9],
    [10, 11, 12]
]

# Matrix addition
result = [
    [A[i][j] + B[i][j] for j in range(3)]
    for i in range(2)
]

print("Resultant Matrix:")
for row in result:
    print(row)
```
