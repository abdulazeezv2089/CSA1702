```python
# Map Coloring using CSP

# Regions and their neighboring regions
graph = {
    'A': ['B', 'C'],
    'B': ['A', 'C', 'D'],
    'C': ['A', 'B', 'D'],
    'D': ['B', 'C']
}

colors = ['Red', 'Green', 'Blue']

# Store assigned colors
assignment = {}


def is_safe(region, color):
    for neighbour in graph[region]:
        if neighbour in assignment and assignment[neighbour] == color:
            return False
    return True


def solve():
    # All regions are colored
    if len(assignment) == len(graph):
        return True

    # Select an uncolored region
    for region in graph:
        if region not in assignment:
            break

    # Try each color
    for color in colors:
        if is_safe(region, color):
            assignment[region] = color

            if solve():
                return True

            # Backtracking
            del assignment[region]

    return False


# Solve CSP
if solve():
    print("Map Coloring Solution:")
    for region, color in assignment.items():
        print(region, "->", color)
else:
    print("No solution exists.")
```
