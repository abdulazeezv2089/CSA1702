```python
# Feed Forward Neural Network

import numpy as np
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

# Input data
X = np.array([
    [0, 0],
    [0, 1],
    [1, 0],
    [1, 1]
])

# Output data (AND operation)
y = np.array([0, 0, 0, 1])

# Create Neural Network
model = Sequential()

# Input and hidden layer
model.add(Dense(4, input_dim=2, activation='relu'))

# Output layer
model.add(Dense(1, activation='sigmoid'))

# Compile the model
model.compile(
    optimizer='adam',
    loss='binary_crossentropy',
    metrics=['accuracy']
)

# Train the model
model.fit(X, y, epochs=500, verbose=0)

# Test the model
predictions = model.predict(X, verbose=0)

print("Predicted Values:")
for i in range(len(X)):
    result = 1 if predictions[i][0] >= 0.5 else 0
    print(X[i], "=>", result)

# Test new input
new_input = np.array([[1, 1]])
prediction = model.predict(new_input, verbose=0)

print("\nInput:", new_input[0])
print("Predicted Output:", 1 if prediction[0][0] >= 0.5 else 0)
```
