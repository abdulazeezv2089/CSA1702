```python
# List Methods: Add, Append, Extend and Delete

numbers = [1, 2, 3]

# 1. Add
numbers.insert(0, 0)
print("After Add:", numbers)

# 2. Append
numbers.append(4)
print("After Append:", numbers)

# 3. Extend
numbers.extend([5, 6])
print("After Extend:", numbers)

# 4. Delete
numbers.remove(3)
print("After Delete:", numbers)
```
